### 1. What types are involved in the following expressions? What data type will each expression evaluate to?

`8 * 2.5`

```python
# Your code here
print("8 is an", type(8))
print("2.5 is an", type(2.5))
print('8 * 2.5 returns a', type(8*2.5))
```

`9 / 2`

```python
print("9 is an", type(9))
print("2 is an", type(2))
print('9 / 2 returns a', type(9/2))
```

`9 // -2`

```python
print("9 is an", type(9))
print("2 is an", type(-2))
print('9 / 2 returns a', type(9//-2))
```

`1.5 * 2 >= 7 - 3`

```python
result = 1.5 * 2 >= 7 - 3
print(f"1.5 * 2 = {1.5 * 2}")
print(f"7 - 3 = {7 - 3}")
print(f"Is {1.5 * 2} >= {7 - 3}? {result}")
```
